```python
import sqlite3
import random
import pandas as pd

# Connect to database (creates file if not exists)
conn = sqlite3.connect("employees.db")
cursor = conn.cursor()

# Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    emp_id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary REAL,
    years_experience INTEGER,
    performance_score REAL
)
""")

# Departments list
departments = ["Engineering", "Sales", "Marketing", "HR", "Finance"]

# Insert synthetic data (40+ records)
data = []
for i in range(1, 51):  # 50 records
    name = f"Employee_{i}"
    dept = random.choice(departments)
    salary = random.randint(50000, 150000)
    exp = random.randint(1, 15)
    score = round(random.uniform(1.0, 5.0), 2)

    data.append((i, name, dept, salary, exp, score))

cursor.executemany("""
INSERT INTO employees (emp_id, name, department, salary, years_experience, performance_score)
VALUES (?, ?, ?, ?, ?, ?)
""", data)

conn.commit()

print("Database created and data inserted successfully!")
```

Task 2: SQL Queries

Query 1

```python
query1 = """
SELECT name, department, salary, performance_score
FROM employees
WHERE performance_score >= 4.0 AND years_experience >= 3
ORDER BY performance_score DESC
LIMIT 15
"""

df_q1 = pd.read_sql_query(query1, conn)
df_q1
```

Query 2

```python
query2 = """
SELECT *
FROM employees
WHERE salary BETWEEN 70000 AND 110000
AND (department = 'Engineering' OR department = 'Sales')
ORDER BY department, salary DESC
"""

df_q2 = pd.read_sql_query(query2, conn)
df_q2
```

Query 3

```python
query3 = """
SELECT department, COUNT(*) AS employee_count, AVG(salary) AS avg_salary
FROM employees
GROUP BY department
ORDER BY avg_salary DESC
"""

df_q3 = pd.read_sql_query(query3, conn)
df_q3
```

Task 3: Pandas Analysis

Average Performance Score by Department

```python
df_all = pd.read_sql_query("SELECT * FROM employees", conn)

avg_perf = df_all.groupby("department")["performance_score"].mean()
avg_perf
```

Replicate Query 1 using Pandas ONLY

```python
df_pandas_q1 = (
    df_all.loc[
        (df_all["performance_score"] >= 4.0) &
        (df_all["years_experience"] >= 3),
        ["name", "department", "salary", "performance_score"]
    ]
    .sort_values(by="performance_score", ascending=False)
    .head(15)
)

df_pandas_q1
```
